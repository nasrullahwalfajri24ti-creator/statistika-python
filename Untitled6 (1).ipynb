```python
import glob

file_excel = glob.glob("*.xlsx")[0]

print("File yang digunakan:", file_excel)
```

```python
import pandas as pd

raw = pd.read_excel(file_excel, header=None)

raw.head(10)
```

```python
data = raw.iloc[7:, 1:8].copy()

data.columns = [
    "ID",
    "Name",
    "Quiz",
    "Mid Term",
    "Final Term",
    "Final Score",
    "Grade"
]

data = data.dropna(subset=["ID"]).reset_index(drop=True)

print(data.head())
```

```python
print("Jumlah mahasiswa:", len(data))
print("Kolom:", data.columns.tolist())
```

```python
data.info()
```

```python
frekuensi_grade = data["Grade"].value_counts().sort_index()
frekuensi_grade
```

```python
tabel_frekuensi = pd.DataFrame({
    "Grade": frekuensi_grade.index,
    "Frekuensi": frekuensi_grade.values
})

tabel_frekuensi
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))

plt.plot(
    frekuensi_grade.index,
    frekuensi_grade.values,
    marker="o",
    linewidth=2
)

plt.title("Distribusi Grade Mahasiswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    frekuensi_grade.index,
    frekuensi_grade.values
)

plt.title("Jumlah Mahasiswa Berdasarkan Grade")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")

plt.show()
```

```python
grade_pie = frekuensi_grade[frekuensi_grade > 0]

plt.figure(figsize=(7,7))

plt.pie(
    grade_pie.values,
    labels=grade_pie.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Persentase Grade Mahasiswa")

plt.show()
```

```python
nilai = data["Final Score"]

nilai
```

```python
nilai = pd.to_numeric(data["Final Score"], errors="coerce").dropna()

print("Mean:", nilai.mean())
print("Median:", nilai.median())
print("Mode:", nilai.mode().iloc[0])
```

```python
print("Minimum:", nilai.min())
print("Maximum:", nilai.max())
```

```python
print("Variance:", nilai.var())
print("Standard Deviation:", nilai.std())
```

```python
print("Skewness:", nilai.skew())
print("Kurtosis:", nilai.kurt())
```

```python
Q1 = nilai.quantile(0.25)
Q2 = nilai.quantile(0.50)
Q3 = nilai.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
se = nilai.std() / (len(nilai) ** 0.5)

print("Standard Error:", se)
```

```python
batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

print("Batas bawah:", batas_bawah)
print("Batas atas:", batas_atas)
```

```python
outlier = data[
    (data["Final Score"] < batas_bawah) |
    (data["Final Score"] > batas_atas)
]

outlier
```
